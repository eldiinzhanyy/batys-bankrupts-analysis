**Чистка**

брнш часть, чистканы бастау, иин/бин норм типка келтирип сондагы пропусктарды алп тастау


```python
df["ИИН/БИН"] = df["ИИН/БИН"].astype(str)
```

```python
df.head()
```

осы жерде жаксы бир проблема пайда болды, флоаттан стр-га ауыстырганда .0 дейтин нарсе пайда болады екен, оны реттеу ушин, ең алдымен интка айналдырып кейин стр га айналдыру керек.


```python
df = df.dropna(subset = ["ИИН/БИН"])
```

```python
df.isna().sum()
```

убрал дубликаты и теперь иин/биндагы дататипты стр га аустрам

```python
df["ИИН/БИН"] = df["ИИН/БИН"].astype(str).str.replace(r"\.0$", "", regex=True)
```

екнш часть, дубликаттарды алам

```python
print("до:", df.shape)
df = df.drop_duplicates(subset="ИИН/БИН").copy()
print("после:", df.shape)
```

третья часть, дата решения суда ны datetime га келтиру

```python
df["Дата решения суда"] = pd.to_datetime(df["Дата решения суда"], errors="coerce", dayfirst=True)
```

```python
df["Дата решения суда"].dtype
```

```python
df["Дата решения суда"].isna().sum()
```

```python
print("всего строк:", len(df))
print("пустых дат:", df["Дата решения суда"].isna().sum())
print("заполнено дат:", df["Дата решения суда"].notna().sum())
```

```python
df["Год решения"] = df["Дата решения суда"].dt.year
```

```python
df.head()
```

финал название компаний дурыстау

```python
df.columns.tolist()
```

```python
df["Наименование  налогоплательщика"] = df["Наименование  налогоплательщика"].str.replace("\n", "").str.strip()
```

```python
df.to_csv("bankrupts_zko_clean.csv", index=False, encoding="utf-8-sig")
```

**Визуализация**

```python
import matplotlib.pyplot as plt
```

```python
by_year = df["Год решения"].value_counts().sort_index()
by_year
```

```python
by_year.plot(kind="bar")
plt.title("Банкротства юрлиц ЗКО по годам")
plt.xlabel("Год")
plt.ylabel("Количество компаний")
plt.show()
```

```python
by_director = df["ФИО руководителя"].value_counts()
by_director
```

```python
serial = by_director[by_director > 1]
serial
```

```python
serial.sort_values().plot(kind="barh")
plt.title("Руководители с несколькими банкротствами")
plt.xlabel("Количество компаний")
plt.show()
```

```python
df["Наименование  налогоплательщика"].str.contains("Товарищество")
```

```python
import numpy as np
```

```python
условия = [
   df["Наименование  налогоплательщика"].str.contains("товарищество|тоо", case=False, na=False),
df["Наименование  налогоплательщика"].str.contains("Индивидуальный|ИП", na=False),
]
ответы = ["ТОО", "ИП"]

df["Форма"] = np.select(условия, ответы, default="Прочее")
```

```python
df["Форма"].value_counts()
```

```python
df[df["Форма"] == "Прочее"]["Наименование  налогоплательщика"]
```

```python
forma.plot(kind="pie", ylabel="", autopct="%1.0f%%")
plt.title("Распределение банкротов по форме")
plt.show()
```
