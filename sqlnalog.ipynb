```python
import pandas as pd
```

```python
df = pd.read_csv('bankrupts_zko_clean.csv')
```

```python
df.head()
```

```python
import sqlite3

conn = sqlite3.connect("bankrupts.db")
df.to_sql("bankrupts", conn, if_exists="replace", index=False)
```

```python
pd.read_sql("SELECT * FROM bankrupts", conn)
```

```python
def sql(запрос):
    return pd.read_sql(запрос, conn)
```

```python
sql("SELECT * FROM bankrupts")
```

```python
sql("""
SELECT "ИИН/БИН", "ФИО руководителя", "Год решения"
FROM bankrupts
""")
```

```python
sql("""
SELECT "ИИН/БИН", "ФИО руководителя", "Год решения"
FROM bankrupts
WHERE "Год решения" = 2024
""")
```

```python
sql("""Select "Год решения",
Count(*) from bankrupts
group by "Год решения" """)
```

```python
sql("""Select "ФИО руководителя",
Count(*) from bankrupts
group by "ФИО руководителя"
order by count(*) desc
limit 10
 """)
```

```python
sql("""Select "ФИО руководителя",
Count(*) from bankrupts
group by "ФИО руководителя"
having count(*) > 1
order by count(*) desc """)
```

```python
sql("""Select "ФИО руководителя",
count(*) from bankrupts
where "ФИО руководителя" is not null
group by "ФИО руководителя"
having count(*) > 1
order by count(*) desc """)
```
